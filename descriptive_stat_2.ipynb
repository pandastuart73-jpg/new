```python
female_students_80_and_above_math = students_data[(students_data['gender'] == 'female') & (students_data['math score'] >= 80)]
female_filtered = female_students_80_and_above_math.drop(columns=['reading score', 'writing score', 'parental level of education', 'lunch', 'test preparation course',"race/ethnicity"]) 
female_filtered
count = len(female_filtered)
print(f"the number of female students who scored 80 and above is {count}")
```

```python
completed_students = students_data[students_data['test preparation course'] =='completed']
completed_students
mean=completed_students["reading score"].mean()
mean
```

```python
students_data["writing score"].max()-students_data["writing score"].min()
```
