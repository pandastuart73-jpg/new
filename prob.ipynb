find the evenT of a number that is less 7

```python
sample_space = {1,2,3,4,5,6}
E = {2,4,6}
A = {x for x in sample_space if x>3}
B= {x for x in sample_space if x<7}

print(f"Event E : {E}")
print(f"Event A: {A}")
print(f"Event B: {B}")
```

WE CAN USE THE FRACTIONS PACKAGE TO AVOID DECIMALS

```python
from fractions import Fraction
```

```python
def prob(event,sample_space):
    event = set(event)& set(sample_space)
    return Fraction(len(event),len(sample_space))
```

```python
## the probability of getting an even number.
print(f"the probability of getting an even number: {prob(E,sample_space)}")
```

```python
print(f"the probability of numbers greater than 3:   {prob(A,sample_space)}")
```

```python
print(f"the probability of numbers less than 7:   {prob(B,sample_space)}")
```

what is the probability of rolling a 7 on a die, of rolling a number less than 10?. Use probablity to check.

```python
seven = {x for x in sample_space if x == 7}
less_than_ten = {x for x in sample_space if x < 10}

print(f"P(rolling a 7): {prob(seven, sample_space)}")
print(f"P(rolling less than 10): {prob(less_than_ten, sample_space)}")
```

```python
from itertools import product
```

```python
three_coins = set(product(['H','T'], repeat=3))
print("3 coins:", three_coins)
print("number of outcomes:", len(three_coins))
```

```python
four_coins = set(product(['H','T'], repeat=4))
print("4 coins:", four_coins)
print("number of outcomes:", len(four_coins))
```

```python
two_dice = set(product([1,2,3,4,5,6], repeat=2))
print("2 dice:", two_dice)
print("number of outcomes:", len(two_dice))
```

## BELOW ARE THE OUTCOMES

- 3 coins: 2^3 = 8 outcomes
- 4 coins: 2^4 = 16 outcomes
- 2 dice: 6^2 = 36 outcomes

### Defining events
An event is a subset of the sample space. For example "exactly 2 heads" is a subset of the 3-coin space.

```python
exactly_two_heads = {c for c in three_coins if c.count('H') == 2}
print("exactly 2 heads:", exactly_two_heads)
print("P(exactly 2 heads):", Fraction(len(exactly_two_heads), len(three_coins)))
```

```python
at_least_one_head = {c for c in three_coins if 'H' in c}
print("P(at least one head):", Fraction(len(at_least_one_head), len(three_coins)))
```

```python
sum_seven = {d for d in two_dice if sum(d) == 7}
print("P(sum = 7):", Fraction(len(sum_seven), len(two_dice)))
```

### Exercise 3.1
1. For 3 coins, find P(at least one head). Do it two ways: (a) directly, (b) using the complement P(E') = 1 - P(not E).

```python
at_least_one_head = {c for c in three_coins if 'H' in c}
print("P(at least one head):", Fraction(len(at_least_one_head), len(three_coins)))
```

```python
no_heads = {c for c in three_coins if 'H' not in c}
print("P(at least one head) using complement:", 1 - Fraction(len(no_heads), len(three_coins)))
```

2. Generate other possible probabilities from the sample spaces of 3 coins, 4 coins and 2 dice.

```python
one_head  = {c for c in three_coins if c.count('H') == 1}
two_heads = {c for c in three_coins if c.count('H') == 2}
all_heads = {c for c in three_coins if c.count('H') == 3}

print("P(exactly 1 head):", Fraction(len(one_head),  len(three_coins)))
print("P(exactly 2 heads):", Fraction(len(two_heads), len(three_coins)))
print("P(all heads):",      Fraction(len(all_heads), len(three_coins)))
```

```python
two_heads_4 = {c for c in four_coins if c.count('H') == 2}
no_heads_4  = {c for c in four_coins if 'H' not in c}

print("P(exactly 2 heads):",  Fraction(len(two_heads_4), len(four_coins)))
print("P(no heads):",         Fraction(len(no_heads_4),  len(four_coins)))
```
