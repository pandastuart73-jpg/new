```python
import pandas as pd
import numpy as np
```

```python
housing = pd.read_csv("Housing.csv")
housing.head()
```

```python
housing.info()
```

```python
event_3_bedrooms = housing['bedrooms'] > 3
event_3_bedrooms.sum()
prefered_outcome = event_3_bedrooms.sum()
sample_space = housing.shape[0]
prob_3_bedrooms = prefered_outcome / sample_space
print(f"The probability is {prob_3_bedrooms:.3f}")
```

alternatively

```python
def prob(event):
    return event.mean()
event_3_bedrooms = housing["bedrooms"] > 3
prob(event_3_bedrooms)
print(f"The probability is {prob_3_bedrooms:.4f}")
```

```python
event_price = housing['price'] > 10000000
price = prob(event_price)
print(f"The probability is {price:.4f}")
```

 example 1;find the probability that a house selected randomly has more than 3 bedrooms and more than 2 bathrooms.

```python
event_bedrooms_and_bathrooms = (housing['bedrooms'] > 3) & (housing['bathrooms'] > 2)
prob = prob(event_bedrooms_and_bathrooms)
print(f"the probability is {prob:.3f}")
```
