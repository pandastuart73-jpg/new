```python
# Sample space for one die
S = {1,2,3,4,5,6}

# Event: die shows a prime number
prime_event = {2,3,5}

# compute probability
def prob(event, sample_space):
    return len(event) / len(sample_space)

print(f"Prime event: {prime_event}")
print(f"P(prime) = {len(prime_event)}/{len(S)} = {prob(prime_event, S)}")
```

```python
import itertools

coin = ['H','T']

def prob(event, sample_space):
    return len(event) / len(sample_space)

# Sample spaces
S3 = set(itertools.product(coin, repeat=3))
S4 = set(itertools.product(coin, repeat=4))
S5 = set(itertools.product(coin, repeat=5))

print(f"3 coins: {len(S3)} outcomes = 2^3")
print(f"4 coins: {len(S4)} outcomes = 2^4")
print(f"5 coins: {len(S5)} outcomes = 2^5\n")
print(S4)
```

```python
# Method (a) Directly - count outcomes with >=1 H
at_least_one_H = { outcome for outcome in S3 if outcome.count('H') >= 1 }
p_direct = prob(at_least_one_H, S3)

print("Q1 (a) Direct:")
print(f"Event at least one head: {at_least_one_H}")
print(f"P(at least one H) = {len(at_least_one_H)}/{len(S3)} = {p_direct}") # 7/8 = 0.875
```

```python
# Method (b) Using complement: P(E) = 1 - P(not E)
# not E = no heads = all Tails
not_E = { outcome for outcome in S3 if outcome.count('H') == 0 } # {('T','T','T')}
p_not_E = prob(not_E, S3)
p_complement = 1 - p_not_E

print("\nQ1 (b) Complement:")
print(f"Complement (no heads): {not_E}")
print(f"P(not E) = {len(not_E)}/{len(S3)} = {p_not_E}")
print(f"P(E) = 1 - P(not E) = 1 - {p_not_E} = {p_complement}")
```

```python
# Q2: Generate any other possible probabilities from sample spaces of 3,4,5
# Bonus - do many

# For 3 coins
exactly_2H_3 = { o for o in S3 if o.count('H') == 2 }
exactly_3H_3 = { o for o in S3 if o.count('H') == 3 }
print(f"\n--- Bonus Probabilities ---")
print(f"3 coins: P(exactly 2H) = {len(exactly_2H_3)}/{len(S3)} = {prob(exactly_2H_3, S3)}")
print(f"3 coins: P(exactly 3H) = {len(exactly_3H_3)}/{len(S3)} = {prob(exactly_3H_3, S3)}")
```

```python

# For 4 coins
exactly_2H_4 = { o for o in S4 if o.count('H') == 2 }
at_least_3H_4 = { o for o in S4 if o.count('H') >= 3 }
print(f"4 coins: P(exactly 2H) = {len(exactly_2H_4)}/{len(S4)} = {prob(exactly_2H_4, S4)}") # 6/16
print(f"4 coins: P(at least 3H) = {len(at_least_3H_4)}/{len(S4)} = {prob(at_least_3H_4, S4)}") # 5/16
```

```python

# For 5 coins
exactly_3H_5 = { o for o in S5 if o.count('H') == 3 }
all_same_5 = { o for o in S5 if o.count('H')==5 or o.count('T')==5 }
print(f"5 coins: P(exactly 3H) = {len(exactly_3H_5)}/{len(S5)} = {prob(exactly_3H_5, S5)}") # 10/32
print(f"5 coins: P(all same) = {len(all_same_5)}/{len(S5)} = {prob(all_same_5, S5)}") # 2/32
```
