```python
import pandas as pd
import numpy as np 
import matplotlib.pylab as plt
import seaborn as sns
```

```python
df = pd.read_csv('StudentsPerformance.csv')
```

```python
df['math score'].plot()
plt.ylabel("math score")
plt.title("plot of math")
plt.show()
```

```python
df.plot (kind= 'hist', x='math score', y='reading score', color='red')
plt.title('scater')
plt.show()
```

```python
fig, axs = plt.subplots(2,3,figsize=(12,10))
axs[0,0].hist(df['math score'],bins=20,color='green')
axs[0,0].set_title('histo')
axs[0,0].set_xlabel('math score')

fig, axs = plt.subplots(2,3,figsize=(12,10))
axs[0,1].hist(df['reading score'],bins=20,color='purple')
axs[0,1].set_title('reading score histogram')
axs[0,1].set_xlabel('reading score')

read_score= df.groupby('reading score')['math score'].mean()
fig, axs = plt.subplots(2,3,figsize=(12,10))
axs[0,2].bar(
    reading_score.index,
    reading_score.values,
    color='lightgreen'
    )
axs[0,2].set_title('average')
axs[0,2].set_xlabel('reading score')
axs[0.2].set_ylabel('frequency')
axs[2,1].bar()
plt.tight_layout()
plt.show
```
